```python
import matplotlib.pyplot as plt
import pandas as pd

# Load dataset
df = pd.read_csv("ecommerce_dataset.csv")

# Convert Order_Date to datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Display dataset info & preview
print("Dataset Overview:")
print(df.info())
df.head()
```

```python
sales_by_date = df.groupby("Order_Date")["Sales"].sum()

plt.figure(figsize=(12, 5))
plt.plot(sales_by_date.index, sales_by_date.values, color="dodgerblue", linewidth=1.5)
plt.title("Total Sales by Date", fontsize=14, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
profit_by_date = df.groupby("Order_Date")["Profit"].sum()

plt.figure(figsize=(12, 5))
plt.plot(profit_by_date.index, profit_by_date.values, color="forestgreen", linewidth=1.5)
plt.title("Total Profit by Date", fontsize=14, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Total Profit ($)")
plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Create YearMonth period and aggregate sales
df["YearMonth"] = df["Order_Date"].dt.to_period("M")
monthly_sales = df.groupby("YearMonth")["Sales"].sum()
monthly_sales.index = monthly_sales.index.astype(str)

highest_month = monthly_sales.idxmax()
highest_sales_val = monthly_sales.max()

plt.figure(figsize=(10, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o", color="navy", linewidth=2, label="Monthly Sales")

# Highlight highest sales month
plt.axvline(x=highest_month, color="crimson", linestyle="--", label=f"Peak Month: {highest_month} (${highest_sales_val:,.2f})")
plt.scatter([highest_month], [highest_sales_val], color="crimson", s=120, zorder=5)

plt.title("Monthly Sales Trend & Peak Month Identification", fontsize=14, fontweight="bold")
plt.xlabel("Month")
plt.ylabel("Total Sales ($)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()

print(f"Highest Sales Month: {highest_month} with Total Sales of ${highest_sales_val:,.2f}")
```

```python
sales_cat = df.groupby("Category")["Sales"].sum()
plt.bar(sales_cat.index, sales_cat.values)
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
profit_city = df.groupby("City")["Profit"].sum()
plt.bar(profit_city.index, profit_city.values)
plt.title("Profit by City")
plt.xlabel("City")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.show()
```

```python
top5 = df.groupby("Product")["Sales"].sum().nlargest(5)
plt.bar(top5.index, top5.values)
plt.title("Top 5 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

```python
for col in ["Sales", "Profit"]:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)][col]
    print(f"Possible Outliers in {col}:")
    print(outliers if not outliers.empty else "No outliers found\n")
```

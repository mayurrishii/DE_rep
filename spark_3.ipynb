```python
import pandas as pd
import numpy as np

df = pd.read_csv("employee_data.csv")

null_cols = df.columns[df.isnull().any()].tolist()
print("Columns with NULL values:", null_cols)


```

```python
mean_age = df["Age"].mean()
print(f"Calculated Mean Age: {mean_age:.2f}")
df["Age"] = df["Age"].fillna(round(mean_age, 1))
print("Remaining NaN count in Age:", df["Age"].isnull().sum())
print("\nFirst 10 rows of Age:")
print(df[["EmployeeID", "Name", "Age"]].head(10))
```

```python
salary_median = df["Salary"].median()
df["Salary"] = df["Salary"].fillna(salary_median)
print(df[["EmployeeID", "Name", "Age","Salary"]].head(10))


```

```python
department_mode = df["Department"].mode()[0]
df["Department"] = df["Department"].fillna(department_mode)
print(df[["EmployeeID", "Name", "Age","Salary", "Department"]].head(10))
```

```python
per_score = df["PerformanceScore"].mean()
df["PerformanceScore"] = df["PerformanceScore"].fillna(per_score)
print(df[["EmployeeID", "Name", "Age","PerformanceScore"]].head(10))

```

```python
# Total count of missing values across the entire DataFrame (should be 0)
print("Total remaining nulls in DataFrame:", df.isnull().sum().sum())
```

```python
# 1. Convert 'JoiningDate' column into datetime format
df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])
# 2. Extract joining year into a new column
df["joining_year"] = df["JoiningDate"].dt.year
# 3. Extract joining month into a new column
df["joining_month"] = df["JoiningDate"].dt.month
# 4. Extract joining day into a new column
df["joining_day"] = df["JoiningDate"].dt.day
# 5. Find how many employees joined after the year 2019
joined_after_2019 = (df["joining_year"] > 2019).sum()
print("Number of employees who joined after 2019:", joined_after_2019)

print(df.head(10))
```

```python
# Split on the '|' delimiter and expand into 2 separate columns
df[["email", "phone"]] = df["Contact"].str.split("|", expand=True)

# Remove leading/trailing spaces from both new columns
df["email"] = df["email"].str.strip()
df["phone"] = df["phone"].str.strip()
print(df[["email","phone"]])
```

```python
df = df.drop(columns=["Contact"])
```

```python
df["bonus"] = df["Salary"] * 0.10
```

```python
df["total_salary"] = df["Salary"] + df["bonus"]
```

```python
df["annual_salary"] = df["total_salary"] * 12
```

```python
df["expafter5years"] = df["Experience"] + 5
```

```python
# Display relevant modified and newly created columns
cols_to_view = ["Name", "Salary", "bonus", "total_salary", "annual_salary", "Experience", "expafter5years", "email", "phone"]

print(df[cols_to_view].head(10))
```

```python
import numpy as np

# 1. Define conditions
conditions = [
    df["Experience"] >= 5,
    (df["Experience"] >= 1) & (df["Experience"] < 5),
    df["Experience"] < 1
]

# 2. Define corresponding labels
choices = ["Senior", "Junior", "Fresher"]

# 3. Create the new column
df["experience_level"] = np.select(conditions, choices, default="Unknown")

# View the result
print(df[["Name", "Experience", "experience_level"]].head(10))
```

```python
import pandas as pd

movies = pd.read_csv(r"movies_data.csv")
movies.head()

```

```python
credits = pd.read_csv(r"credits_data.csv")
credits.head()
```

```python
movies = movies.merge(credits, on = 'title')

movies.head()
```

```python
movies.shape
```

```python
movies.columns
```

```python
movies = movies[['movie_id', 'title', 'overview', 'genres', 'keywords', 'cast', 'crew']]
```

```python
movies.head()
```

```python
import ast

def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name'])
    return L
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies.head()
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies.head()
```

```python
def convert3(text):
    l = []
    counter = 0
    for i in ast.literal_eval(text):
        if counter < 3:
            l.append(i['name'])
        counter += 1
    return l

movies['cast'] = movies['cast'].apply(convert3)
```

```python
def fetch_dir(text):
    l = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            l.append(i['name'])
    return l

movies['crew'] = movies['crew'].apply(fetch_dir)

# movies.sample(5)
```

```python
movies.head()
```

```python
def collapse(l):
    l1 = []
    for i in l:
        l1.append(i.replace(" ", ""))
    return l1
```

```python
movies['cast']= movies['cast'].apply(collapse)
movies['crew']= movies['crew'].apply(collapse)
movies['genres']= movies['genres'].apply(collapse)
movies['keywords']= movies['keywords'].apply(collapse)
```

```python
movies['overview'] = movies['overview'].apply(lambda x: x.split() if isinstance(x,str) else [])
```

```python
movies['tags']= movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
new = movies.drop(columns = ['overview', 'genres', 'keywords', 'cast', 'crew'])
```

```python
new.head()
```

```python
new['tags'] = new['tags'].apply(lambda x: " ".join(x) if isinstance(x, list) else x)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features = 5000, stop_words = 'english')
```

```python
cv
```

```python
vector = cv.fit_transform(new['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity.shape
```

```python
def recommend(movie):
    # Get the index of the given movie
    index = new[new['title'] == movie].index[0]
    
    # Calculate top 5 most similar movies
    distances = sorted(list(enumerate(similarity[index])), reverse=True, key=lambda x: x[1])
    
    # Print the top 5 recommended movie titles (excluding the movie itself at index 0)
    for i in distances[1:6]:
        print(new.iloc[i[0]].title)
```

```python
recommend('Titanic')
```

```python
import requests, time

API_KEY = "6f5063ef2e5f56af3a5f04aed5082430"

def fetch_poster(movie_id):
    url = f"https://api.themoviedb.org/3/movie/{movie_id}?api_key={API_KEY}&language=en-US"
    try:
        response = requests.get(url, timeout=10)   # add timeout
        response.raise_for_status()                # raise HTTPError if not 200
        data = response.json()
        poster_path = data.get('poster_path')
        if poster_path:
            return "https://image.tmdb.org/t/p/w500/" + poster_path
    except requests.exceptions.RequestException as e:
        print(f" Could not fetch poster for movie_id={movie_id}: {e}")
    return None

def recommend(movie):
    index = new[new['title'] == movie].index[0]
    distances = sorted(list(enumerate(similarity[index])), reverse=True, key=lambda x: x[1])

    recommended_movies = []
    recommended_posters = []

    for i in distances[1:6]:
        movie_id = new.iloc[i[0]].movie_id
        recommended_movies.append(new.iloc[i[0]].title)
        poster = fetch_poster(movie_id)
        recommended_posters.append(poster if poster else "No poster available")

    return recommended_movies, recommended_posters
```

```python
movies, posters = recommend("Avengers: Age of Ultron")
for title, poster in zip(movies, posters):
    print(title, ":", poster)
```

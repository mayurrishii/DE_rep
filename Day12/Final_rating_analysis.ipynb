```python
from kafka import KafkaConsumer
import json
from collections import defaultdict

consumer = KafkaConsumer(
    'orders_topic',
    'drivers_topic',
    'feedback_topic',
    bootstrap_servers='localhost:9092',
    auto_offset_reset='earliest',
    group_id='analytics-group',
    value_deserializer=lambda x: json.loads(x.decode('utf-8'))
)

restaurant_sales = defaultdict(int)
restaurant_ratings = defaultdict(list)
city_orders = defaultdict(int)

print("Real-Time Analytics Started...\n")

for message in consumer:

    topic = message.topic
    data = message.value

    print("\nReceived From:", topic)
    print(data)

    # orders analysis
    if topic == "orders_topic":

        restaurant_sales[data['restaurant']] += data['price']
        city_orders[data['city']] += 1

        print("\n LIVE SALES REPORT ")

        for k, v in restaurant_sales.items():
            print(k, ":", v)

        print("\n CITY ORDER COUNT ")

        for k, v in city_orders.items():
            print(k, ":", v)

    # feedback analysis
    elif topic == "feedback_topic":

        restaurant_ratings[data['restaurant']].append(data['rating'])

        print("\n live rating analysis")

        for restaurant, ratings in restaurant_ratings.items():

            avg_rating = sum(ratings) / len(ratings)

            print(
                restaurant,
                "Average Rating:",
                round(avg_rating, 2)
            )

    # driver analysis
    elif topic == "drivers_topic":

        print("\nDRIVER STATUS ")

        print(
            data['name'],
            "in",
            data['location'],
            "-",
            data['status']
        )
```
